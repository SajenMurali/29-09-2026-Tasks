### Student Marks Array

```python
import numpy as np

marks = np.array([78, 65, 89, 56, 92])

print("Marks:", marks)
print("Shape:", marks.shape)
print("Size:", marks.size)
print("Dimension:", marks.ndim)
print("Data type:", marks.dtype)
```

### Student Marks Access

```python
import numpy as np

marks = np.array([72, 85, 64, 90, 76])

print("Marks:", marks)

# Indexing
print("First student:", marks[0])
print("Third student:", marks[2])
print("Last student:", marks[-1])

# Slicing
print("First three students:", marks[:3])
print("Students 2 to 4:", marks[1:4])
print("Last two students:", marks[-2:])
```

### Subject-wise Marks

```python
import numpy as np

marks = np.array([
    78, 85, 90,
    65, 72, 80,
    88, 91, 84,
    56, 62, 70,
    95, 89, 92
])

marks = marks.reshape(5, 3)

print("Student-wise marks:")
print(marks)

print("Shape:", marks.shape)
```

### Internal and External Marks

```python
import numpy as np

internal = np.array([18, 16, 19, 15, 17])
external = np.array([70, 65, 72, 60, 75])

final_marks = internal + external

print("Internal marks:", internal)
print("External marks:", external)
print("Final marks:", final_marks)
```

### Pass Percentage Analysis

```python
import numpy as np

marks = np.array([45, 78, 56, 32, 91])

passed = marks[marks >= 50]

print("Marks:", marks)
print("Students scoring 50 or above:", passed)
```

### Average Marks

```python
import numpy as np

marks = np.array([
    [78, 85, 90],
    [65, 72, 80],
    [88, 91, 84],
    [56, 62, 70],
    [95, 89, 92]
])

average = np.mean(marks, axis=1)

print("Marks:")
print(marks)

print("Average marks of each student:")
print(average)
```

### Class Performance Statistics

```python
import numpy as np

marks = np.array([67, 82, 91, 74, 58])

print("Marks:", marks)

print("Total:", np.sum(marks))
print("Average:", np.mean(marks))
print("Highest:", np.max(marks))
print("Lowest:", np.min(marks))
print("Standard deviation:", np.std(marks))
```

### Subject-wise Performance

```python
import numpy as np

marks = np.array([
    [78, 85, 90],
    [65, 72, 80],
    [88, 91, 84],
    [56, 62, 70],
    [95, 89, 92]
])

subject_total = np.sum(marks, axis=0)

print("Marks:")
print(marks)

print("Total marks in each subject:")
print(subject_total)
```

### Student-wise Performance

```python
import numpy as np

marks = np.array([
    [78, 85, 90],
    [65, 72, 80],
    [88, 91, 84],
    [56, 62, 70],
    [95, 89, 92]
])

student_total = np.sum(marks, axis=1)

print("Marks:")
print(marks)

print("Total marks of each student:")
print(student_total)
```

### Student Ranking

```python
import numpy as np

marks = np.array([245, 278, 219, 290, 256])
sorted_marks = np.sort(marks)[::-1]
print("Original marks:", marks)
print("Marks in descending order:", sorted_marks)
ranking = np.argsort(marks)[::-1]

print("Student ranking:")
for rank, student_index in enumerate(ranking, start=1):
    print("Rank", rank, "- Student", student_index + 1,
          "- Marks:", marks[student_index])
    
```

### Duplicate Marks Analysis

```python
import numpy as np

marks = np.array([85, 92, 85, 76, 92])

unique_marks = np.unique(marks)

print("Original marks:", marks)
print("Unique marks:", unique_marks)
```

### Missing Marks

```python
import numpy as np

marks = np.array([78, 85, np.nan, 92, 67])

average = np.nanmean(marks)

print("Marks:", marks)
print("Average excluding missing value:", average)
```

### Grade Classification

```python
import numpy as np

marks = np.array([95, 82, 74, 61, 45])

grades = np.select(
    [
        marks >= 90,
        marks >= 80,
        marks >= 70,
        marks >= 60
    ],
    [
        "A",
        "B",
        "C",
        "D"
    ],
    default="F"
)

print("Marks:", marks)
print("Grades:", grades)
```

### Random Marks Generation

```python
import numpy as np
marks = np.random.randint(0, 101, 5)

print("Randomly generated marks:", marks)

print("Total:", np.sum(marks))
print("Average:", np.mean(marks))
print("Highest:", np.max(marks))
print("Lowest:", np.min(marks))
print("Standard deviation:", np.std(marks))
```
